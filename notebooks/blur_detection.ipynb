# Blur Detection

This notebook demonstrates how to detect blur in images using the variance of the Laplacian method.

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Load test images
sharp_image = cv2.imread("../data/test_images/car.jpg")
sharp_image = cv2.cvtColor(sharp_image, cv2.COLOR_BGR2RGB)

# Create a blurred version for testing
blurred_image = cv2.GaussianBlur(sharp_image, (15, 15), 0)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 5))
ax1.imshow(sharp_image)
ax1.set_title("Sharp Image")
ax2.imshow(blurred_image)
ax2.set_title("Blurred Image")
plt.show()
```

```python
# Calculate variance of Laplacian for blur detection
def detect_blur(image, threshold=100):
    # Convert to grayscale
    gray = cv2.cvtColor(image, cv2.COLOR_RGB2GRAY)

    # Compute the Laplacian of the image
    laplacian = cv2.Laplacian(gray, cv2.CV_64F)

    # Compute the variance
    variance = laplacian.var()

    # Check if image is blurry
    is_blurry = variance < threshold

    return variance, is_blurry

# Test on sharp image
sharp_variance, sharp_blurry = detect_blur(sharp_image)
print(f"Sharp image variance: {sharp_variance:.2f}, Is blurry: {sharp_blurry}")

# Test on blurred image
blurred_variance, blurred_blurry = detect_blur(blurred_image)
print(f"Blurred image variance: {blurred_variance:.2f}, Is blurry: {blurred_blurry}")

# Visualize the Laplacian
gray_sharp = cv2.cvtColor(sharp_image, cv2.COLOR_RGB2GRAY)
laplacian_sharp = cv2.Laplacian(gray_sharp, cv2.CV_64F)

gray_blurred = cv2.cvtColor(blurred_image, cv2.COLOR_RGB2GRAY)
laplacian_blurred = cv2.Laplacian(gray_blurred, cv2.CV_64F)

fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(10, 8))
ax1.imshow(gray_sharp, cmap='gray')
ax1.set_title("Sharp Grayscale")
ax2.imshow(gray_blurred, cmap='gray')
ax2.set_title("Blurred Grayscale")
ax3.imshow(laplacian_sharp, cmap='gray')
ax3.set_title(f"Laplacian (var: {sharp_variance:.2f})")
ax4.imshow(laplacian_blurred, cmap='gray')
ax4.set_title(f"Laplacian (var: {blurred_variance:.2f})")
plt.show()
```
